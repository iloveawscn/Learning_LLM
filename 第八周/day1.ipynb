# 设置 modal tokens

## 重要 - 请务必阅读并遵循这些说明！

首先请访问：https://modal.com

注册一个账户。然后点击右上角的头像菜单，选择 "Settings"

然后点击左侧边栏的 "API Tokens"，再点击 "New Token"。

你会得到类似这样的命令来运行：

`modal token set --token-id ak-somethinghere --token-secret as-somethinghere`


### 故障排除

如果有问题，可以尝试方法：


来自 Windows的建议：

> 如果其他使用 Windows 的人遇到此问题：除了必须从命令提示符运行 `modal token new` 外，还需要移动生成的 token 文件。它会将 token 文件（.modal.toml）部署到你的 Windows 配置文件夹。虚拟环境无法看到该位置（奇怪的是，即使我设置了环境变量并重启后也无法看到）。我将该 token 文件移动到我操作实验的文件夹后，就不再出现认证错误了。

手动方式：

直接将 2 个密钥添加到你的 .env 文件中可能完全没问题：

```
MODAL_TOKEN_ID=ak-...
MODAL_TOKEN_SECRET=as-...
```

然后重新运行 `load_dotenv(override=True)` 来加载这些环境变量。

# 在我们继续之前 -

## 我们需要将你的 HuggingFace Token 设置为 Modal 中的密钥

## 非常重要 - 请阅读 - 这让很多人困惑！

Modal 中的密钥有一个描述该密钥的**名称**。  
然后密钥本身有一个 KEY 和一个 VALUE。  
我们将设置一个密钥：  

名称：huggingface-secret  
密钥：HF_TOKEN  
值：hf_...  

## 万无一失的方法：

1. 访问 modal.com，登录并进入你的仪表板  
2. 点击导航栏中的 Secrets  
3. 创建新密钥，点击 Hugging Face，这个新密钥需要命名为 **huggingface-secret**，因为我们在代码中就是这样引用的  
4. 填写你的密钥为 HF_TOKEN，值为你实际的 token hf_...  
5. 点击完成

# 可选：保持 Modal 温暖

## 提高 Modal pricer 服务速度的方法

第一次运行此 modal 类时，构建可能需要长达 10 分钟。  
之后应该会快得多.. 如果需要唤醒则 30 秒，否则 2 秒。  
如果你想让它始终保持 2 秒，可以通过编辑 pricer_service2.py 中的这个常量来防止容器休眠：

`MIN_CONTAINERS = 0`



将其改为 1 以保持容器存活。  
但请注意：这会消耗积分！只有在你愿意让进程持续运行的情况下才这样做。

或者，你可以运行此代码，它将保持温暖 20 分钟而不是 2 分钟。

### 保持温暖 20 分钟后冷却的代码：

